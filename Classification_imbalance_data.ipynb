```python
df = df.drop(columns=["EmployeeCount", "StandardHours", "Over18", "EmployeeNumber"])
```

```python
df.shape
```

```python
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
from sklearn.model_selection import train_test_split

X=df.drop("Attrition", axis=1)
Y=df["Attrition"]

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42, stratify=Y)
```

```python
y_test.shape
```

```python
numeric_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
```

```python
categorical_cols = X.select_dtypes(include=["str"]).columns.tolist()
```

```python
print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
```

```python
df["Education"].unique()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing  import OneHotEncoder, StandardScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer
scaling_pipe=Pipeline([
    ("scaling", StandardScaler())
])

onehot_pipe = Pipeline([
    ("onehot", OneHotEncoder(sparse_output=False))
])
ordinal_pipe = Pipeline([
     ("ordinal", OrdinalEncoder(categories = [['Non-Travel','Travel_Rarely', 'Travel_Frequently']]))
 ])

preprocessing = ColumnTransformer([
    ("scaling", scaling_pipe, numeric_cols),
    ("onehot", onehot_pipe, ['Department', 'EducationField', 'Gender', 'JobRole', 'MaritalStatus', 'OverTime']),
    ("ordinal", ordinal_pipe, ["BusinessTravel"] )
])

```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
full_pipe_lr = Pipeline([
    ("preprocessing", preprocessing),
    ("model_lr", LogisticRegression(random_state=42, max_iter=1000, class_weight='balanced'))
])

full_pipe_rf = Pipeline([
    ("preprocessing", preprocessing),
    ("model_rf", RandomForestClassifier(random_state=42, n_jobs=-1, class_weight='balanced'))
])

full_pipe_xgb = Pipeline([
    ("preprocessing", preprocessing),
    ("model_xgb", XGBClassifier(random_state=42, n_jobs=-1, scale_pos_weight=(len(y_train) - sum(y_train)) / sum(y_train)))
])
```

```python
param_grid_lr = {
    "model_lr__C": [0.01, 0.1, 1, 10]
}

param_grid_rf = {
    "model_rf__n_estimators": [100, 200],
    "model_rf__max_depth": [None, 10]
}

param_grid_xgb = {
    "model_xgb__n_estimators": [100, 200],
    "model_xgb__max_depth": [3, 5],
    "model_xgb__learning_rate": [0.1, 0.01]
}
```

```python
from sklearn.model_selection import GridSearchCV

grid_lr = GridSearchCV(full_pipe_lr, param_grid_lr, cv=5, scoring="roc_auc")
grid_rf = GridSearchCV(full_pipe_rf, param_grid_rf, cv=5, scoring="roc_auc")
grid_xgb = GridSearchCV(full_pipe_xgb, param_grid_xgb, cv=5, scoring="roc_auc")
```

```python
grid_lr.fit(x_train, y_train)
grid_rf.fit(x_train, y_train)
grid_xgb.fit(x_train, y_train)

print("LR best score:", grid_lr.best_score_)
print("RF best score:", grid_rf.best_score_)
print("XGB best score:", grid_xgb.best_score_)
```

```python
y_pred_lr = grid_lr.best_estimator_.predict(x_test)

from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(y_test, y_pred_lr))
```
