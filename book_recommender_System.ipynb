# **Popularity Based Recommender System**


```python
rating_with_name = rating.merge(book, on="ISBN")
rating_with_name
```

```python
num_rating_df = rating_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating' : 'No_rating'},inplace= True)
num_rating_df
```

```python
avg_rating = rating_with_name.groupby('Book-Title')['Book-Rating'].mean().reset_index()
avg_rating.rename(columns={'Book-Rating' : 'avg_rating'}, inplace= True)
avg_rating
```

```python
popular_book = num_rating_df.merge(avg_rating, on= "Book-Title")
popular_book
```

```python
popular_df=popular_book[popular_book['No_rating'] >= 250].sort_values('avg_rating', ascending=False).head(50)
popular_df
```

```python
popular_df.merge(book, on='Book-Title').drop_duplicates('Book-Title')[['Book-Title', 'Book-Author', 'Publisher','Image-URL-L' ,'No_rating', 'avg_rating']]
```

# **Collabrative Filtering  Based recommender System**

```python
x = rating_with_name.groupby('User-ID').count()['Book-Rating'] > 200
max_rating_user = x[x].index
```

```python
filtered_rating = rating_with_name[rating_with_name['User-ID'].isin(max_rating_user)]
```

```python
filtered_rating
```

```python
y = filtered_rating.groupby('Book-Title').count()['Book-Rating'] >= 50
famous_books = y[y].index
```

```python
famous_books
```

```python
final_ratings =  filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
pt = final_ratings.pivot_table(index='Book-Title', columns='User-ID', values='Book-Rating')
```

```python
pt.fillna(0, inplace=True)
```

```python
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_score = cosine_similarity(pt)
```

```python
similarity_score.shape
```

```python
def recommend(book_name):
  index = np.where(pt.index == book_name)[0][0]
  distance = similarity_score[index]
  similar_items = sorted(list(enumerate(similarity_score[index])), key= lambda x :x[1], reverse = True)[1:6]

  for i in similar_items:
    print(pt.index[i[0]])
```

```python
recommend('2nd Chance')
```
